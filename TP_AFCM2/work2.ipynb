```python
import pandas as pd
import numpy as np
```

```python
# load data2
data2 = pd.read_csv('data2.csv')

data2
```

```python
# add IDs to the data2

data2['ID'] = np.arange(1, len(data2) + 1)

data2.to_csv('data2.csv', index=False)
```

```python
# put the ID column in the first position
cols = data2.columns.tolist()
cols = cols[-1:] + cols[:-1]
data2 = data2[cols]

#save data2
data2.to_csv('data2.csv', index=False)

data2
```

```python
# count the number of unique values in each column
data = pd.read_csv('data2.csv')

data.nunique()

# get count of each value in all columns
for col in data.columns[1:]:
    print(data[col].value_counts())

```

```python
# one hot encode data_encoded , except the ID and city column
data_encoded = pd.get_dummies(data, columns=data.columns[2:])

#save data_encoded

data_encoded.to_csv('data_encoded.csv', index=False)
```
