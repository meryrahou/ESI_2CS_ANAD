```python
import pandas as pd
import numpy as np
```

```python
# load data1
data1 = pd.read_csv('data1.csv')

data1
```

```python
# Convert text to uppercase in column: 'Current school (or last school attended)'
data1['Current school (or last school attended)'] = data1['Current school (or last school attended)'].str.upper()
# save data1
data1.to_csv('data1.csv', index=False)

data1
```

```python
# add IDs to the data1

data1['ID'] = np.arange(1, len(data1) + 1)

data1.to_csv('data1.csv', index=False)
```

```python
# put the ID column in the first position
cols = data1.columns.tolist()
cols = cols[-1:] + cols[:-1]
data1 = data1[cols]

#save data1
data1.to_csv('data1.csv', index=False)

data1
```

```python
# capitalize City, turn every alger into algiers in that column
# in On what device(s) will you use DataCamp?, turn every My own mobile iOS/Android device into My own mobile device

data1['City'] = data1['City'].str.capitalize()
data1['City'] = data1['City'].replace('Alger', 'Algiers')

data1['On what device(s) will you use DataCamp?'] = data1['On what device(s) will you use DataCamp?'].replace('My own mobile iOS/Android device', 'My own mobile device')

# save data1
data1.to_csv('data1.csv', index=False)

data1
```

```python
# one hot encode every column except the ID column
data1 = pd.get_dummies(data1, columns=data1.columns[1:-1])

data1.to_csv('data_encoded.csv', index=False)
```
